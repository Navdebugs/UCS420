```python
# QUESTION 1
roll_no = input("Enter your roll number: ").strip()

digits = [int(d) for d in roll_no]
L = [digit * 10 for digit in digits]

print("\ni. Initial list L:")
print(L)

L.append(500)
print("\nAfter append(500):")
print(L)

L.insert(2, 250)
print("\nAfter insert(250, 2):")
print(L)
# append() adds the new element at the end of the list.
# insert() adds the new element at the specific position.

value_to_remove = L[0]
L.remove(value_to_remove)
print("\nAfter remove():")
print(L)

removed_element = L.pop()
print("\nAfter pop():")
print(L)

L.sort()
print("\nAscending order:")
print(L)

L.sort(reverse=True)
print("\nDescending order:")
print(L)

print("\nFirst three elements:")
print(L[:3])

print("Last three elements:")
print(L[-3:])

average = sum(L) / len(L)
greater_than_average = [x for x in L if x > average]

print("\nAverage of L:", average)
print("Elements greater than average:")
print(greater_than_average)
```

```python
# QUESTION 2
scores = tuple(L[:8])
print("\nScores tuple:")
print(scores)

highest_score = max(scores)
highest_index = scores.index(highest_score)

lowest_score = min(scores)
lowest_count = scores.count(lowest_score)

print("\nHighest score:", highest_score)
print("Index of highest score:", highest_index)
print("Lowest score:", lowest_score)
print("Number of times lowest score appears:", lowest_count)

reversed_scores = list(reversed(scores))

print("\nReversed tuple as list:")
print(reversed_scores)
# Tuples are immutable, so their elements cannot be changed or reversed in place.

search_score = int(input("\nEnter a score to search: "))

if search_score in scores:
    print("First occurrence index:", scores.index(search_score))
else:
    print("Score is not present in the tuple.")


print("\nAttempting to modify tuple:")
try:
    scores[0] = 100
except TypeError as e:
    print("Error:", e)

first_score, second_score, *remaining_scores = scores

print("\nFirst score:", first_score)
print("Second score:", second_score)
print("Remaining scores:", remaining_scores)
```

```python
# QUESTION 3 : RANDOM NUMBERS

import random
from collections import Counter

roll_no = input("Enter your roll number: ").strip()
random.seed(int(roll_no))

random_numbers = [random.randint(100, 900) for _ in range(100)]

print("\nGenerated random numbers:")
print(random_numbers)

def is_prime(n):
    if n < 2:
        return False

    for i in range(2, int(n ** 0.5) + 1):
        if n % i == 0:
            return False
    return True


# Q3(ii)
odd_numbers = [x for x in random_numbers if x % 2 != 0]

print("\nOdd numbers:")
print(odd_numbers)
print("Number of odd numbers:", len(odd_numbers))


# Q3(iii)
even_numbers = [x for x in random_numbers if x % 2 == 0]

print("\nEven numbers:")
print(even_numbers)
print("Number of even numbers:", len(even_numbers))


# Q3(iv)
prime_numbers = [x for x in random_numbers if is_prime(x)]

print("\nPrime numbers:")
print(prime_numbers)
print("Number of prime numbers:", len(prime_numbers))


# Q3(v)
frequency = Counter(random_numbers)
most_frequent_number, highest_frequency = frequency.most_common(1)[0]

print("\nMost frequently occurring number:", most_frequent_number)
print("Number of times it occurs:", highest_frequency)
```

```python
# QUESTION 4
roll_no = input("Enter your roll number: ").strip()

digits = [int(d) for d in roll_no]
first_8_digits = digits[:8]

A = {digit * 7 for digit in first_8_digits}
B = {digit * 9 for digit in first_8_digits}

print("\nSet A:")
print(A)

print("\nSet B:")
print(B)

# Q4(vi) UNION
union_set = A.union(B)

print("\nUnion of A and B:")
print(union_set)

# Q4(vii) INTERSECTION
intersection_set = A.intersection(B)

print("\nIntersection of A and B:")
print(intersection_set)

# Q4(viii) Find A-B and B-A separately
A_minus_B = A.difference(B)
B_minus_A = B.difference(A)

print("\nA - B:")
print(A_minus_B)

print("B - A:")
print(B_minus_A)

# difference() gives elements unique to one set, while symmetric_difference() gives elements unique to either set.

# Q4(ix) Find the symmetric difference of A and B
symmetric_difference = A.symmetric_difference(B)

print("\nSymmetric difference of A and B:")
print(symmetric_difference)

# Q4(x) Check whether A is a subset of B and B is a superset of A
print("\nIs A a subset of B?")
print(A.issubset(B))

print("Is B a superset of A?")
print(B.issuperset(A))

# Q4(xi) Remove X from A using discard()
X = int(input("\nEnter a value X to remove from set A: "))

A.discard(X)

print("Set A after discard():")
print(A)

# discard() is safer because it does not give an error if X is not present in the set.
```

```python
# QUESTION 5

my_dict = {
    "name": "navdeep jaim",
    "roll_no": "1024170082",
    "branch": "CSE",
    "age": 20,
    "city": "jalandhar"
}

print("Original dictionary:")
print(my_dict)


# Q5(i)
my_dict["location"] = my_dict.pop("city")

print("\nAfter renaming city to location:")
print(my_dict)

# Q5(ii)
my_dict["cgpa"] = 7.39

print("\nAfter adding CGPA:")
print(my_dict)

# Q5(iii)
my_dict["age"] += 1

print("\nAfter increasing age by 1:")
print(my_dict)

# Q5(iv) Delete "branch" using pop() and del
dict_using_pop = my_dict.copy()
dict_using_del = my_dict.copy()

removed_value = dict_using_pop.pop("branch")

print("\nAfter using pop():")
print(dict_using_pop)
print("Removed value:", removed_value)

del dict_using_del["branch"]

print("\nAfter using del:")
print(dict_using_del)

# pop() removes the key and returns its value, while del removes the key without returning the value.


# Q5(v)
print("\nKey-value pairs:")

for key, value in my_dict.items():
    print(f"{key} → {value}")


# Q5(vi)
print("\nChecking for email:")

if "email" in my_dict:
    print("Email:", my_dict["email"])
else:
    print("Email key is not present in the dictionary.")


# Q5(vii)
friend_dict = {
    "name": "Rahul",
    "roll_no": "1024170350",
    "branch": "CSE",
    "age": 21,
    "city": "Delhi"
}

merged_dict = {**my_dict, **friend_dict}

print("\nFriend dictionary:")
print(friend_dict)

print("\nMerged dictionary:")
print(merged_dict)

# When the same key exists in both dictionaries, the value from the second dictionary wins.


# Q5(viii)
string_values_dict = {
    key: value
    for key, value in my_dict.items()
    if isinstance(value, str)
}

print("\nDictionary containing only string values:")
print(string_values_dict)
```
