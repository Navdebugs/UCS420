```python
# Assignment 5.3
n = int(input("Enter n: "))

s = 0

for num in range(2, n + 1):
    prime = True

    for i in range(2, num):
        if num % i == 0:
            prime = False
            break

    if prime:
        s = s + num

print("Sum of prime numbers =", s)
```

```python
# Assignment 6.1

def AddOdd(n):
    s = 0

    for i in range(1, n + 1):
        if i % 2 != 0:
            s = s + i

    return s


n = int(input("Enter n: "))

print("Sum of odd numbers =", AddOdd(n))
```

```python
# Assignment 6.2

def IsPrime(n):
    if n < 2:
        return False

    for i in range(2, n):
        if n % i == 0:
            return False

    return True


def AddPrime(n):
    s = 0

    for i in range(2, n + 1):
        if IsPrime(i):
            s = s + i

    return s


n = int(input("Enter n: "))

print("Sum of prime numbers =", AddPrime(n))
```
