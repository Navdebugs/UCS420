```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd

M = float(input("Enter the value of M: "))

x = np.linspace(-10, 10, 100)

y1 = M * x**2
y2 = M * np.sin(x)

plt.figure(figsize=(8, 5))

plt.plot(x, y1, color="blue", linestyle="-", label="M × x²")
plt.plot(x, y2, color="red", linestyle="--", label="M × sin(x)")

plt.xlabel("x values")
plt.ylabel("y values")
plt.title("Comparison of Two Mathematical Functions")
plt.grid(True)
plt.legend()

plt.show()
```

```python
subjects = ["Maths", "Python", "Statistics", "DBMS", "English"]
scores = [85, 92, 78, 88, 75]

data = pd.DataFrame({
    "Subject": subjects,
    "Score": scores
})

print(data)

plt.figure(figsize=(8, 5))

colors = ["skyblue", "orange", "green", "purple", "red"]

bars = sns.barplot(
    x="Subject",
    y="Score",
    data=data,
    hue="Subject",
    palette=colors,
    legend=False
)

for bar in bars.patches:
    height = bar.get_height()
    plt.text(
        bar.get_x() + bar.get_width() / 2,
        height + 1,
        int(height),
        ha="center"
    )

plt.title("Subject-wise Scores")
plt.xlabel("Subjects")
plt.ylabel("Marks")
plt.grid(axis="y")

plt.show()
```

```python
np.random.seed(1024170082)

values = np.random.randint(1, 101, 50)

print("Generated values:")
print(values)

cumulative_values = np.cumsum(values)

plt.figure(figsize=(10, 5))

plt.subplot(1, 2, 1)
plt.plot(
    cumulative_values,
    color="blue",
    linestyle="-",
    marker="o"
)
plt.title("Cumulative Sum")
plt.xlabel("Position")
plt.ylabel("Cumulative Value")
plt.grid(True)

noise = np.random.normal(0, 5, 50)

plt.subplot(1, 2, 2)
plt.scatter(
    values,
    noise,
    color="red"
)
plt.title("Random Noise Scatter Plot")
plt.xlabel("Generated Values")
plt.ylabel("Noise")
plt.grid(True)

plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("company_sales_data.csv")

display(df)
```

```python
plt.figure(figsize=(8, 5))

sns.lineplot(
    x=df["month_number"],
    y=df["total_profit"],
    marker="o",
    color="blue"
)

plt.title("Monthly Profit")
plt.xlabel("Month")
plt.ylabel("Total Profit")
plt.grid(True)

plt.show()
```

```python
products = [
    "facecream",
    "facewash",
    "toothpaste",
    "bathingsoap",
    "shampoo",
    "moisturizer"
]

plt.figure(figsize=(12, 6))

for product in products:
    sns.lineplot(
        x=df["month_number"],
        y=df[product],
        marker="o",
        label=product
    )

plt.title("Product Sales Over the Year")
plt.xlabel("Month")
plt.ylabel("Number of Units Sold")
plt.grid(True)
plt.legend()
plt.show()
```

```python
product_totals = df[products].sum()

plt.figure(figsize=(10, 6))

sns.barplot(
    x=product_totals.index,
    y=product_totals.values,
    color="skyblue"
)

plt.title("Total Sales of Each Product")
plt.xlabel("Products")
plt.ylabel("Total Units Sold")
plt.xticks(rotation=30)
plt.grid(axis="y")

plt.show()
```

```python
plt.figure(figsize=(10, 6))

sns.barplot(
    x=df["month_number"],
    y=df["total_profit"],
    color="lightgreen"
)

plt.title("Monthly Total Profit")
plt.xlabel("Month")
plt.ylabel("Total Profit")
plt.grid(axis="y")

plt.show()
```
